# STRINGS

## 1.String Concatenation

```python
'''Write a Python program that takes two strings i.e string 1 “Hello ”, string 2 get name as
input from the user and concatenates them together. Display the concatenated string as
the output.'''
    
str1 = "Hello"
str2 = input("Enter your name:")
msg  = str1 + " " + str2
print(msg)
```

```python
'''Now concatenate string 3 “, welcome to Python programming” to the existing string and
display the output string.'''

str3 = ", Welcome to Python Programming"
msg2 = msg + " " + str3
print(msg2)
```

## 2.String Slicing and Indexing

```python
# Print the first character of the string.
msg2[0]
```

```python
# Print the last character of the string.
len(msg2)
msg2[41]
```

```python
# Print the first 5 characters of the string.
msg2[:5]
```

```python
# Print the last 11 characters of the string.
msg2[-11:]
```

```python
# Print the string in reverse.
msg2[::-1]
```

```python
# Use slicing and print the word “Python” from the existing string.
msg2[24:30]
```

## 3.String Methods

```python
strM = "Python beginner tutorial"
print(strM)
```

```python
# Convert the sentence to uppercase.
print(strM.upper())
```

```python
# Convert the sentence to lowercase.
print(strM.lower())
```

```python
# Use Capitalize and return the sentence to the original input form.
print(strM.capitalize())
```

```python
# Count the total number of occurrences of character ‘t’ in the string.
print(strM.count('t'))
```

```python
# Replace all occurrences of “Python” with “Machine Learning” in the input string
# strM = “Python beginner tutorial"
strM.replace("Python","Machine Learning")
```

# TUPLES

```python
# Create 1st tuple with values -> (10, 20, 30), 2nd tuple with values -> (40, 50, 60):
tuple1 = (10,20,30)
tuple2 = (40,50,60)
print(tuple1,tuple2)
```

```python
# Concatenate the two tuples and store it in “t_combine”)
t_combine=tuple1+tuple2
print(t_combine)
```

```python
# Repeat the elements of “t_combine” 3 times
newtuple=(t_combine) * 3
print(newtuple)
```

```python
# Access the 3rd element from “t_combine”
t_combine[2]
```

```python
# Access the first three elements from “t_combine”
t_combine[:3]
```

```python
# Access the last three elements from “t_combine”
t_combine[-3:]
```
