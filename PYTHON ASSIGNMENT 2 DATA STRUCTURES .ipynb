```python
# a. Append the string "Yazhini" to name_list

name_list = ["Anju","Diya","Swati","Sheba","Arya"]
name_list.append("Yazhini")
name_list
```

## **Sets (Operations)**

```python
# a. Create a set called my_set with following values:
#           ['a','e','i','o','u','a','a','i']

my_set = {'a','e','i','o','u','a','a','i'} # duplicate values gets removed!
my_set                            
```

```python
# b. Attempt to change the value of my_set[4] = 's'
#If code throws an error, provide an explanation

my_set[4] = 's'
my_set

'''In Python, sets are unordered collections of unique elements. 
They do not support indexing, so we cannot access or modify an element using an index like my_set[4].'''
```

```python
my_set.add('s')
print(my_set)  #New value can be added like this at a random index
```

```python
# c. Create two sets:
#          set1 with values: {1, 3, 5, 7, 9}
#          set2 with values: {2, 3, 5, 8, 10}

set1 = {1,3,5,7,9}
set2 = {2, 3, 5, 8, 10}

print(set1)
print(set2)
```

```python
# d. Compute and print the union and intersection of set1 and set2.

#Union
set3 = set1.union(set2) 
set3
```

```python
#Intersection

set3 = set1.intersection(set2) 
set3
```

**Performance Category Program**

```python
#1. Prompt user for Input. Score range should be from 0 to 10 (both inclusive).

score = int(input("Enter your score (0 to 10): "))

if score < 0 or score > 10:
    print("Invalid score! Please enter a score between 0 and 10.")
else:
    print("Valid score:", score)
```

```python
# 2. Find the performance category based on the input score using following criteria:
# a. Above Average: Score greater than 7
# b. Average: Score between 4 and 7(both inclusive)
# c. Below Average: Score lesser than 4

score = int(input("Enter your score (0 to 10): "))

if score>7:
    print("Performance Category: Above Average")
    print("Excellent work! Keep up your great performance.")
elif score>=4:
    print("Performance Category: Average")
    print("Good effort! Keep practicing to improve your performance.")
else:
    print("Performance Category: Below Average")
    print("Need to improve your performance.Practice more for better results.")
    
```
